```python
from dataclasses import dataclass, field
from typing import Dict, Any

def make_dataclass_with_dict_attributes(**kwargs):
    fields = {}
    for name, value in kwargs.items():
        fields[name] = (Dict[str, Any], field(default_factory=dict))

    @dataclass
    class DynamicDataClass:
        __annotations__ = fields

        def __post_init__(self):
            for name, value in self.__dict__.items():
                if isinstance(value, dict):
                    for key, val in value.items():
                        setattr(self, f"{name}_{key}", val)

    return DynamicDataClass

# Example usage
DynamicDataClass = make_dataclass_with_dict_attributes(
    attribute1=dict,
    attribute2=dict,
    attribute3=dict
)

# Create an instance of the dataclass
instance = DynamicDataClass(
    attribute1={"key1": "value1", "key2": "value2"},
    attribute2={"keyA": "valueA", "keyB": "valueB"},
    attribute3={"keyX": "valueX", "keyY": "valueY"}
)

# Access the attributes
print(instance.attribute1_key1)  # Output: value1
print(instance.attribute2_keyA)  # Output: valueA
print(instance.attribute3_keyX)  # Output: valueX
```
